```python
import time, subprocess
from pathlib import Path
_cwd = Path.cwd().resolve()
REPO = next(
    (p for p in (_cwd, *_cwd.parents) if (p / 'lo7_corpus.test.yaml').is_file()),
    _cwd.parent,
)
out = REPO / '_lo7' / 'nb_manifest.json'
out.parent.mkdir(parents=True, exist_ok=True)
t0 = time.perf_counter()
subprocess.run(
    ['uv', 'run', 'lo7-manifest', '--config', str(REPO / 'lo7_corpus.test.yaml'),
     '--ref-date', '2019-02-15', '--out', str(out)],
    cwd=REPO, check=True,
)
print(f'snappiness_wall_s: {time.perf_counter() - t0:.4f}')
```

## Transfiguration hooks

- **Zoom**: slice `heatmap.cell_dates` / `cell_levels` to a sub-window (in-memory) or write `derived.json` from manifest only.
- **Maximize / minimize**: UI chrome only; same HTML.
- **Multimodal export**: static slide deck or screenshot from `lo7-heatmap` output; optional `--motion` for §9 display rhythm.

Next cell writes `nb_heatmap.html` from the manifest (no repo walk).

```python
import subprocess
from pathlib import Path
_cwd = Path.cwd().resolve()
REPO = next(
    (p for p in (_cwd, *_cwd.parents) if (p / 'lo7_corpus.test.yaml').is_file()),
    _cwd.parent,
)
subprocess.run(
    ['uv', 'run', 'lo7-heatmap', '--in', str(REPO / '_lo7' / 'nb_manifest.json'),
     '--out', str(REPO / '_lo7' / 'nb_heatmap.html')],
    cwd=REPO, check=True,
)
print('wrote', REPO / '_lo7' / 'nb_heatmap.html')
```
